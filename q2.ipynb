```python
from collections import deque

def get_successors(state, cap_a=4, cap_b=3):
    a, b = state
    successors = []

    # 1. Fill Jug A
    successors.append(((cap_a, b), "Fill Jug A (4L)"))
    # 2. Fill Jug B
    successors.append(((a, cap_b), "Fill Jug B (3L)"))
    # 3. Empty Jug A
    successors.append(((0, b), "Empty Jug A"))
    # 4. Empty Jug B
    successors.append(((a, 0), "Empty Jug B"))

    # 5. Pour A -> B
    pour_a_to_b = min(a, cap_b - b)
    successors.append(((a - pour_a_to_b, b + pour_a_to_b), "Pour Jug A into Jug B"))

    # 6. Pour B -> A
    pour_b_to_a = min(b, cap_a - a)
    successors.append(((a + pour_b_to_a, b - pour_b_to_a), "Pour Jug B into Jug A"))

    return successors

def solve_water_jug():
    # Initial state: both jugs are empty
    start_state = (0, 0)
    queue = deque([[start_state, []]])
    visited = {start_state}

    while queue:
        (a, b), path = queue.popleft()

        # Goal check: either jug has exactly 2 litres
        if a == 2 or b == 2:
            return path + [((a, b), "Goal reached!")]

        for next_state, action in get_successors((a, b)):
            if next_state not in visited:
                visited.add(next_state)
                queue.append([next_state, path + [((a, b), action)]])

    return None

path = solve_water_jug()
if path:
    print("Minimum sequence of steps to measure exactly 2 litres:")
    for i, (state, action) in enumerate(path):
        print(f"Step {i}: {action:25} -> State: (Jug A: {state[0]}L, Jug B: {state[1]}L)")
else:
    print("No solution found.")
```
