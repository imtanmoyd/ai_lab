```python
def dfs_has_route(graph, start, destination, visited=None):
    if visited is None:
        visited = set()

    if start == destination:
        return True

    visited.add(start)

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            if dfs_has_route(graph, neighbor, destination, visited):
                return True
    return False

def get_connected_components(graph):
    visited = set()
    components = []

    # Inner helper to explore a component via DFS
    def dfs_explore(node, component):
        visited.add(node)
        component.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                dfs_explore(neighbor, component)

    for node in graph:
        if node not in visited:
            component = []
            dfs_explore(node, component)
            components.append(component)

    return components

# --- Example Road Network ---
# We have three independent groups of cities:
# Group 1: New York, Boston, Philadelphia
# Group 2: Los Angeles, San Francisco, San Diego
# Group 3: Chicago (isolated)
road_network = {
    'New York': ['Boston', 'Philadelphia'],
    'Boston': ['New York'],
    'Philadelphia': ['New York'],
    'Los Angeles': ['San Francisco', 'San Diego'],
    'San Francisco': ['Los Angeles'],
    'San Diego': ['Los Angeles'],
    'Chicago': []
}

# Task 1: Check routes
city_a, city_b = 'New York', 'Philadelphia'
city_c, city_d = 'New York', 'Los Angeles'

print(f"Route exists between '{city_a}' and '{city_b}': {dfs_has_route(road_network, city_a, city_b)}")
print(f"Route exists between '{city_c}' and '{city_d}': {dfs_has_route(road_network, city_c, city_d)}")
print("-" * 50)

# Task 2: Find all connected components
components = get_connected_components(road_network)
print(f"Total connected components found: {len(components)}")
for idx, component in enumerate(components, 1):
    print(f"Group {idx}: {', '.join(component)}")
```
